**Бинарный поиск**

```python
from random import randint
def gen_seq(n):
  seq = []
  for i in range(n):
    a = randint(0, n)
    seq.append(a)
  return seq

def sort_seq(list_1):
    for i in range(len(list_1)):
      for j in range(len(list_1) - 1):
        if list_1[j] > list_1[j+1]:
          list_1[j], list_1[j+1] = list_1[j+1], list_1[j]
    return list_1


def binary_search(seq, n):
  min_index = 0
  max_index = len(seq)-1
  result = -1
  while min_index < max_index:
    mid_index = (min_index + max_index) // 2
    if seq[mid_index] == n:
      result = mid_index
      max_index = mid_index - 1
    elif seq[mid_index] > n:
      max_index = mid_index - 1
    else:
      min_index = mid_index + 1
  if result != -1:
    return (result, seq[result])
  else:
    return (0, 0)

L = gen_seq(10)
L_sorted = sort_seq(L)
a = binary_search(L_sorted, 4)
print(a)
print(L_sorted)

```
