```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

a = factorial(5)
print(a)
```

```python
def gcd(num_1, num_2):
  if num_2 == 0 or num_1 == 0:
    return num_1
  return gcd(num_2, num_1 % num_2)

gcd(25, 5)
```

```python
from random import randint

def gen_seq():
    seq = []
    for i in range(10):
        a = randint(0, 11)
        seq.append(a)
    return seq

def sort_seq(list_1):
    for i in range(len(list_1)):
        for j in range(len(list_1) - 1):
            if list_1[j] > list_1[j+1]:
                list_1[j], list_1[j+1] = list_1[j+1], list_1[j]
    return list_1  # ДОБАВИТЬ ВОЗВРАТ!

def binary_search(seq, key, min_index=0, max_index=None):
    if max_index is None:
        max_index = len(seq) - 1

    if min_index > max_index:
        return -1

    mid_index = (min_index + max_index) // 2

    if seq[mid_index] == key:
        if mid_index == 0 or seq[mid_index - 1] != key:
            return mid_index
        else:
            return binary_search(seq, key, min_index, mid_index - 1)
    elif seq[mid_index] > key:
        return binary_search(seq, key, min_index, mid_index - 1)
    else:
        return binary_search(seq, key, mid_index + 1, max_index)

def find_range(first_index, seq):
    if first_index == -1:
        return (-1, -1)

    key = seq[first_index]
    last_index = first_index

    while last_index + 1 < len(seq) and seq[last_index + 1] == key:
        last_index += 1

    return (first_index, last_index)


L = gen_seq()
L = sort_seq(L)
print(L)
key_found = binary_search(L, 4)
print(key_found)
range_tuple = find_range(key_found, L)
print(range_tuple)
```

```python
list_1 = [1, 2, 3, 3, 4, 8, 9,]
list_2 = [0, 2, 2, 4, 4, 9, 10, 11]

def merge(list_1, list_2):
  merged_list = []
  i = j = 0

  while i < len(list_1) and j < len(list_2):
      if list_1[i] <= list_2[j]:
          merged_list.append(list_1[i])
          i += 1
      else:
          merged_list.append(list_2[j])
          j += 1

  while i < len(list_1):
      merged_list.append(list_1[i])
      i += 1

  while j < len(list_2):
      merged_list.append(list_2[j])
      j += 1

  return merged_list

merge(list_1, list_2)
```
