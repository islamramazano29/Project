**Рекурсивные функции**

*Вычисление факториала*


```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

a = factorial(5)
print(a)
```

```python
def gcd(num_1, num_2):
  if num_2 == 0 or num_1 == 0:
    return num_1
  return gcd(num_2, num_1 % num_2)

gcd(25, 5)
```
