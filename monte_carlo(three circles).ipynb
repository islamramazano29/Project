**Монте Карло**

```python
import random

R = 1
N_total = 1000000
count = 0

for i in range(N_total):
  cx = random.choice([-1, 0, 1])

  while True:
    x = random.uniform(cx - R, cx + R)
    y = random.uniform(-R, R)
    if (x - cx)**2 + y**2 <= R*R:
      break


  if x*x + y*y <= R*R and (x+1)**2 + y*y > R*R and (x-1)**2 + y*y > R*R:
    count += 1

S = (count / N_total) * (3 * 3.14159 * R * R)
print(S)
```
